```python
import subprocess
import time
import urllib.request

log = open("server.log", "w")
server = subprocess.Popen(["python", "web/app.py", "--port", "5000"], stdout=log, stderr=subprocess.STDOUT)

print("Loading the model, this takes a few minutes...")
while True:
    try:
        urllib.request.urlopen("http://127.0.0.1:5000/api/health")
        break
    except OSError:
        if server.poll() is not None:
            print(open("server.log").read())
            raise RuntimeError("The app stopped, see the log above.")
        time.sleep(5)
print("Ready.")
```
